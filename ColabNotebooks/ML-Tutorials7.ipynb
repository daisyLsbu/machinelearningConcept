# Tutorial: 7 Feature selection
 

```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
from sklearn.model_selection import KFold, train_test_split
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

import matplotlib.pyplot as plt
```

```python
filename = 'ADNI_Tadpole_data_tutorial.xlsx'
data_df = pd.read_excel(filename)
data_df.head()
```

```python
y_target = data_df['Amyloid status']
x_data = data_df.drop(['PTID', 'SUVR', 'Amyloid status', 'Diagnosis'], axis = 1)
x_train, x_test, y_train, y_test = train_test_split(x_data, y_target, train_size=0.75)
```

```python
clf = LogisticRegression(penalty='none', class_weight='none', max_iter=10000, solver='saga' )
clf.fit(x_train[['ApoE4']], y_train)

y_hat = clf.predict(x_test[['ApoE4']])
y_prob = clf.predict_proba(x_test[['ApoE4']])
print(accuracy_score(y_test, y_hat>0.6))

clf = LogisticRegression(penalty='none', class_weight='none', max_iter=10000, solver='saga' )
clf.fit(x_train[['AGE']], y_train)

y_hat = clf.predict(x_test[['AGE']])
y_prob = clf.predict_proba(x_test[['AGE']])
print(accuracy_score(y_test, y_hat>0.6))

clf = LogisticRegression(penalty='none', class_weight='none', max_iter=10000, solver='saga' )
clf.fit(x_train[['GENDER']], y_train)

y_hat = clf.predict(x_test[['GENDER']])
y_prob = clf.predict_proba(x_test[['GENDER']])
print(accuracy_score(y_test, y_hat>0.6))

clf = LogisticRegression(penalty='none', class_weight='none', max_iter=10000, solver='saga' )
clf.fit(x_train[['EDUCATION']], y_train)

y_hat = clf.predict(x_test[['EDUCATION']])
y_prob = clf.predict_proba(x_test[['EDUCATION']])
print(accuracy_score(y_test, y_hat>0.6))

clf = LogisticRegression(penalty='none', class_weight='none', max_iter=10000, solver='saga' )
clf.fit(x_train[['CDRSB']], y_train)

y_hat = clf.predict(x_test[['CDRSB']])
y_prob = clf.predict_proba(x_test[['CDRSB']])
print(accuracy_score(y_test, y_hat>0.6))

clf = LogisticRegression(penalty='none', class_weight='none', max_iter=10000, solver='saga' )
clf.fit(x_train[['ADAS11']], y_train)

y_hat = clf.predict(x_test[['ADAS11']])
y_prob = clf.predict_proba(x_test[['ADAS11']])
print(accuracy_score(y_test, y_hat>0.6))

clf = LogisticRegression(penalty='none', class_weight='none', max_iter=10000, solver='saga' )
clf.fit(x_train[['ADAS13']], y_train)

y_hat = clf.predict(x_test[['ADAS13']])
y_prob = clf.predict_proba(x_test[['ADAS13']])
print(accuracy_score(y_test, y_hat>0.6))

clf = LogisticRegression(penalty='none', class_weight='none', max_iter=10000, solver='saga' )
clf.fit(x_train[['MMSE']], y_train)

y_hat = clf.predict(x_test[['MMSE']])
y_prob = clf.predict_proba(x_test[['MMSE']])
print(accuracy_score(y_test, y_hat>0.6))

clf = LogisticRegression(penalty='none', class_weight='none', max_iter=10000, solver='saga' )
clf.fit(x_train[['RAVLT_immediate']], y_train)

y_hat = clf.predict(x_test[['RAVLT_immediate']])
y_prob = clf.predict_proba(x_test[['RAVLT_immediate']])
print(accuracy_score(y_test, y_hat>0.6))
```
