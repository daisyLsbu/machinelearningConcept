```python
import csv
```

```python
filename = 'data\Tutorial_3_data.csv'
header = []
data = []
```

```python
csvfile = open(filename)
content_of_file = csv.reader(csvfile)
```

```python
header = next(content_of_file)
for row in content_of_file:
  data.append(row)
```

```python
csvfile.close()
```

```python
print(len(data))
print(type(data))
```

```python
valueofx = float(x[4])
valueofy = float(y[4])
print(valueofx)
print(valueofy)

dataofx = float(data[0][4])
dataofy = float(data[1][4])
print(type(dataofx))
print(type(dataofy))

print(valueofx * valueofy)
print(dataofx * dataofy)
```
